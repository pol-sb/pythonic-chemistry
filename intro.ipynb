## Known issues

- `03-data-analysis/data-analysis-numpy.ipynb` expects `array_example.dat`, which is not currently in the repository. If you hit a `FileNotFoundError` there, skip that cell or create a small space-separated array file with that name.
- `07-bures-chemical-kinetics/bures-method-solved.ipynb` has two paths that omit the `datasets/` prefix. Use the project notebook, or fix those two reads to `datasets/vtna_ex1_init_vals.dat` and `datasets/vtna_ex1_all.dat`.

If a notebook fails on a missing file, check that you opened it from this page rather than moving it out of its module folder.
